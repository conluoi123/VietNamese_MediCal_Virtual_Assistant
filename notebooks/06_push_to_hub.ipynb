```python
import os
from transformers import AutoModelForSeq2SeqLM, T5Tokenizer
from huggingface_hub import login

# 1. Đăng nhập hệ thống bằng token write
HF_TOKEN = ""
if not HF_TOKEN:
    raise ValueError("Vui lòng cung cấp token write của bạn để đăng nhập Hugging Face Hub.")
login(token=HF_TOKEN)

# 2. Kiểm tra nếu đang chạy trên Google Colab
if 'google.colab' in str(get_ipython()):
    print(" Đang chạy trên Google Colab. Tiến hành mount Drive...")
    from google.colab import drive
    drive.mount('/content/drive')
    
    # Đường dẫn chuẩn hóa trên Drive chung của nhóm 
    BEST_MODEL_PATH = "/content/drive/MyDrive/VietNamese_MediCal_Virtual_Assistant/checkpoints/vit5/best"
else:
    print(" Đang chạy trên máy Local (Jupyter Notebook / VS Code)...")
    # Lấy thư mục gốc của project (đi lùi ra ngoài 1 cấp từ thư mục 'notebooks')
    BASE_DIR = os.path.dirname(os.getcwd())
    BEST_MODEL_PATH = os.path.join(BASE_DIR, "checkpoints", "vit5", "best")

print(f" Đường dẫn checkpoint được sử dụng: {BEST_MODEL_PATH}")
# =====================================================================

print("\n Đang nạp mô hình và cấu hình gốc...")
# Giữ use_fast=False để tránh lỗi đồng bộ mã hóa
tokenizer = T5Tokenizer.from_pretrained(BEST_MODEL_PATH, use_fast=False)
model = AutoModelForSeq2SeqLM.from_pretrained(BEST_MODEL_PATH)

# 3. Định danh kho chứa 
HF_USERNAME = "ntthanh0307"  
REPO_ID = f"{HF_USERNAME}/vit5-vimedaq-medical-qa"

print(f"\n Đang đẩy toàn bộ trọng số lên Hub: https://huggingface.co/{REPO_ID}")

# Tiến hành đẩy mã và cấu hình sạch lên Hugging Face Hub
tokenizer.push_to_hub(REPO_ID, private=False, legacy=False)
model.push_to_hub(REPO_ID, private=False)

print("\n XONG! Mô hình từ Drive đã bay lên Hugging Face Hub an toàn và sạch sẽ.")
```
